```python
import re
import os

FILE_NAME = "users.txt"

# -------- Username Validation ----------------
def is_valid_username(username):
    # Regex pattern to validate email format
    # Must start with a letter, contain @ and .
    pattern = r'^[a-zA-Z][a-zA-Z0-9._]*@[a-zA-Z]+\.[a-zA-Z.]+$'

    # Check pattern match and ensure ".@" is not present
    if re.match(pattern, username):
        if ".@" not in username:
            return True
    return False


# -------- Password Validation ----------------
def is_valid_password(password):
    # Check password length
    if len(password) < 6 or len(password) > 16:
        return False

    # Check for at least one uppercase letter
    if not re.search(r"[A-Z]", password):
        return False

    # Check for at least one lowercase letter
    if not re.search(r"[a-z]", password):
        return False

    # Check for at least one digit
    if not re.search(r"[0-9]", password):
        return False

    # Check for at least one special character
    if not re.search(r"[@$!%*?&]", password):
        return False

    return True


# -------- Register User ----------------
def register():
    # Take username input
    username = input("Enter email: ")

    # Validate username
    if not is_valid_username(username):
        print("Invalid username format")
        return

    # Take password input
    password = input("Enter password: ")

    # Validate password
    if not is_valid_password(password):
        print("Invalid password format")
        return

    # Store username and password in file
    with open(FILE_NAME, "a") as file:
        file.write(f"{username}:{password}\n")

    print("Registration successful")


# -------- Login User ----------------
def login():
    # Check if user file exists
    if not os.path.exists(FILE_NAME):
        print("No users found. Please register first.")
        return

    # Take login credentials
    username = input("Enter email: ")
    password = input("Enter password: ")

    # Read file and verify credentials
    with open(FILE_NAME, "r") as file:
        for line in file:
            stored_user, stored_pass = line.strip().split(":")
            if username == stored_user and password == stored_pass:
                print("Login successful")
                return

    # If login fails
    print("Invalid credentials")
    choice = input("Forgot password? (yes/no): ").lower()
    if choice == "yes":
        forgot_password()


# -------- Forgot Password ----------------
def forgot_password():
    # Take username for password reset
    username = input("Enter registered email: ")
    found = False

    # Read all existing user data
    with open(FILE_NAME, "r") as file:
        lines = file.readlines()

    # Rewrite file with updated password
    with open(FILE_NAME, "w") as file:
        for line in lines:
            stored_user, stored_pass = line.strip().split(":")
            if stored_user == username:
                found = True
                new_pass = input("Enter new password: ")

                # Validate new password
                if is_valid_password(new_pass):
                    file.write(f"{stored_user}:{new_pass}\n")
                    print("Password updated successfully")
                else:
                    print("Invalid password format")
                    file.write(line)
            else:
                file.write(line)

    # If username not found
    if not found:
        print("User not found. Please register.")


# ---------------- Main Menu ----------------
while True:
    print("\n1. Register\n2. Login\n3. Exit")
    choice = input("Enter choice: ")

    # Call respective functions based on user choice
    if choice == "1":
        register()
    elif choice == "2":
        login()
    elif choice == "3":
        print("Exiting program")
        break
    else:
        print("Invalid choice")
```
